```python
import pandas as pd
df = pd.read_excel("Dataset for Data Analytics.xlsx")
df
```

```python
df['CouponCode']=df['CouponCode'].fillna('No_Coupon')
print(df['CouponCode'].isnull().sum())
print(df['CouponCode'].value_counts())
```

```python
df.rename({"ItemsInCart":"Items_In_Cart"})
df.head()
```

```python
df
```

```python
df.info()
```

```python
cat_cols = ['OrderStatus', 'Product', 'PaymentMethod', 'ReferralSource', 'CouponCode']

for col in cat_cols:
    print(f"{col}: {df[col].nunique()} unique → {df[col].unique()}")
```

```python
for col in cat_cols:
    df[col] = df[col].astype('category')
print(df[cat_cols].dtypes)
```

```python
 print(df.memory_usage(deep=True).sum() / 1024, "KB")
```

```python
df['UnitPrice'] = df['UnitPrice'].round(2)
df['TotalPrice'] = df['TotalPrice'].round(2)
```

```python
df['calculated'] = (df['Quantity'] * df['UnitPrice']).round(2)
df['discount_rate'] = ((df['calculated'] - df['TotalPrice']) / df['calculated']).round(4)

print(df.groupby('CouponCode')['discount_rate'].unique())
```

```python
df.drop(columns=['calculated', 'discount_rate'], inplace=True)
```

```python
str_cols = ['OrderID', 'CustomerID', 'Product', 'ShippingAddress',
            'PaymentMethod', 'OrderStatus', 'TrackingNumber',
            'ReferralSource', 'CouponCode']

for col in str_cols:
    df[col] = df[col].str.strip()

# spot check a few categorical ones for casing issues
for col in ['OrderStatus', 'Product', 'PaymentMethod']:
    print(f"{col}: {sorted(df[col].unique())}")
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

print(df[['Date', 'Year', 'Month']].head())
```

```python
df
```

```python
df.insert(2, 'Year', df.pop('Year'))
df.insert(3, 'Month', df.pop('Month'))

print(df.columns.tolist())
```

```python
df
```

```python
df['ShippingAddress'] = df['ShippingAddress'].str.replace(' Main St', '', regex=False)
df['ShippingAddress'].head()
```

```python
df
```

```python
df['CustomerID'] = df['CustomerID'].str.replace('C', '', regex=False)
```

```python
df['OrderID'] = df['OrderID'].str.replace('ORD', '', regex=False)
```

```python
df['TrackingNumber'] = df['TrackingNumber'].str.replace('TRK', '', regex=False)
```

```python
df.to_excel("cleaned_dataset.xlsx", index=False)
```
